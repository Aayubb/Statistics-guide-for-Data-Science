## Descriptive Stastics

#### *Prerequisites*
- none


### What are descriptive Statistics?
- They are numers used to summarize and descrive data.
- Descriptive stats are just descriptive meaning they do not involve in generalizing beyond the data at hand.
- Descriptive stats are also *suggestive*.

> **suggestive:** Insights, patterns that the dataset hints at without providing any proof.


#### Here are a list of methods used for descriptive statistcs in python


| Category | Meaning | Examples |
| :------------------ | :-------------------: | -------------------: |
| Central tendendy    | Where is the middle of the data    |**mean(), median(), mode()**|
| Spread/ variability| How spread is the data          |   **variance, std, range, IQR** |
| Shape    | How the distribution looks like             |          **skewness, kurtosis** |
